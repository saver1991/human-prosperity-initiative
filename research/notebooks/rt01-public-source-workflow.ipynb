This optional notebook uses Python **standard library only** to invoke the canonical Node.js 22 CLI and read the source-controlled public build reports. Run from the HPI repository root **after** `npm run check`. It makes no network calls and has no precomputed output cells.

```python
import json
import subprocess
from pathlib import Path

def run_node(script):
    result = subprocess.run(['node', script], check=True, capture_output=True, text=True)
    return json.loads(result.stdout)

verification = run_node('scripts/rt01-pipeline-verify.mjs')
assert verification['status'] == 'PASS'
assert verification['syntheticOnly'] is True
assert verification['causeOrPopulationImpactMeasured'] is False
print('Verified source-based public bundle; no scientific validation claimed')
```
